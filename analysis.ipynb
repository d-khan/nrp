```python
import plotly.express as px
import plotly.graph_objects as go
from scipy.stats import linregress

fig = px.scatter(
    df,
    x="calories",
    y="sodium",
    color="Sodium Level",
    color_discrete_map={
        "2,300 mg or Below": "royalblue",
        "Above 2,300 mg": "red"
    },
    title="Relationship Between Calories and Sodium",
    labels={
        "calories": "Calories (kcal)",
        "sodium": "Sodium (mg)"
    }
)

# Overall regression using ALL observations
slope, intercept, r_value, p_value, std_err = linregress(
    df["calories"],
    df["sodium"]
)

x = [df["calories"].min(), df["calories"].max()]
y = [slope * i + intercept for i in x]

fig.add_trace(
    go.Scatter(
        x=x,
        y=y,
        mode="lines",
        name="Linear Regression",
        line=dict(color="black", width=3)
    )
)
fig.add_hline(
    y=2300,
    line_dash="dash",
    line_color="red",
    annotation_text="Recommended Daily Sodium Limit (2,300 mg)",
    annotation_position="top left"
)

fig.show()
```
